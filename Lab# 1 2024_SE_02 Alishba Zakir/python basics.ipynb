## Exercise 1: Variables and Data Types

```python
# Exercise 1: Variables and Data Types[cite: 1]
width = 640
height = 480
image_name = "sample.png"
is_color = True

print(f"width: {width}, type: {type(width)}")
print(f"height: {height}, type: {type(height)}")
print(f"image_name: {image_name}, type: {type(image_name)}")
print(f"is_color: {is_color}, type: {type(is_color)}")

total_pixels = width * height
print(f"Total number of pixels: {total_pixels}")
```

## Exercise 2: if, elif, and else

```python
# Exercise 2: if, elif, and else[cite: 1]
intensities = [-1, 0, 127, 128, 255, 256]

for val in intensities:
    if val < 0 or val > 255:
        label = "Invalid"
    elif val >= 128:
        label = "Bright pixel"
    else:
        label = "Dark pixel"
    print(f"Intensity {val}: {label}")
```

## Exercise 3: for and while Loops

```python
# Exercise 3: for and while Loops[cite: 1]
# For loop using threshold 128
pixel_values = [20, 130, 255]
threshold = 128

print("--- For Loop Output ---")
for val in pixel_values:
    if val >= threshold:
        label = "Bright pixel"
    else:
        label = "Dark pixel"
    print(f"Pixel {val}: {label}")

# While loop printing 0, 1, and 2
print("\n--- While Loop Output ---")
counter = 0
while counter < 3:
    print(counter)
    counter += 1
```

## Exercise 4: Functions

```python
# Exercise 4: Functions[cite: 1]
def is_bright(pixel, threshold=128):
    return pixel >= threshold

print("is_bright(127):", is_bright(127))
print("is_bright(128):", is_bright(128))
print("is_bright(200, 210):", is_bright(200, 210))
```

## Exercise 5: Lists and Tuples

```python
# Exercise 5: Lists and Tuples[cite: 1]
pixels = [0, 128, 255]
print("First item:", pixels[0])

pixels[1] = 150
pixels.append(64)
print("Updated list:", pixels)
print("Length of list:", len(pixels))

size = (640, 480)
width, height = size
print("Width:", width)
print("Height:", height)

# Explanation: 
# Lists are mutable, meaning their elements can be changed, replaced, or modified after creation. 
# Tuples are immutable, meaning once defined, their elements cannot be modified or replaced in place.
```

## Exercise 6: Dictionaries and Sets

```python
# Exercise 6: Dictionaries and Sets[cite: 1]
image = {"width": 640, "height": 480}
image["mode"] = "RGB"

print("Width:", image["width"])
print("Mode:", image["mode"])

labels = ["cat", "dog", "cat", "bird"]
unique_labels = set(labels)
print("Unique labels set:", unique_labels)
print("Length of unique labels set:", len(unique_labels))
```
